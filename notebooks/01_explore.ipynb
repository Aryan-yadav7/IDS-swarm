```python
import pandas as pd

train = pd.read_csv("../data/UNSW_NB15_training-set.csv")
test = pd.read_csv("../data/UNSW_NB15_testing-set.csv")

print(train.shape, test.shape)
```

```python
print(set(train.columns) ^ set(test.columns))      # empty set = identical columns
print(train.select_dtypes(exclude="number").columns.tolist())
```

```python
print(train.isna().sum().sum(), test.isna().sum().sum())
print((train["service"] == "-").mean())
```

```python
print(train["label"].value_counts(normalize=True))
print(test["label"].value_counts(normalize=True))
```

```python
print(train["attack_cat"].value_counts())
print(test["attack_cat"].value_counts())
```

```python
for col in ["proto", "service", "state"]:
    tr, te = set(train[col]), set(test[col])
    print(col, len(tr), len(te), "unseen in train:", te - tr)
```

```python
print(pd.crosstab(train["attack_cat"], train["label"]))
```
